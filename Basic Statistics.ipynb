*Task1: Load the patient dataset and show summary with info()*

*Task 5: Remove duplicate patient records based on PatientID*

```python
df_patient_unique=df_patient.drop_duplicates('PatientID')
```

```python
df_patient_unique
```

*Task 6: Fill missing BillAmount values with the mean bill amount*

```python
mean_bill_amount=round(df_patient_unique['BillAmount'].mean(),0) #calculating the mean of bill amount and rounding off as a whole number
```

```python
print(f'Mean Bill Amount:{mean_bill_amount}')
```

```python
df_patient_unique['BillAmount'].fillna(mean_bill_amount, inplace=True) #filling the missing bill amount values with the mean
```

```python
df_patient_unique
```

*Task 7: Merge the billing dataset with patient dataset on PatientID*

```python
df_billing.head()
```

*Task 8: Concatenate an additional DataFrame that contains new patients for the current week (row-wise).*

```python
new_patients={
    'PatientID': [106, 107,108],
    'Name': ['Frank', 'Grace','Brendan'],
    'Department': ['Pediatrics', 'Oncology','Cardiology'],
    'Doctor': ['Dr. White', 'Dr. Green','Dr. Smith'],
    'BillAmount': [4500.0, 8000.0,5550]
}       #creating the data for new patients for the current week
```

```python
df_new_patients=pd.DataFrame(new_patients)
```

```python
df_new_patients
```

```python
df_patients_updated=pd.concat([df_patient_unique,df_new_patients],axis=0,ignore_index=True) #concatinating new patients data into the unique data (excluding the duplicate values)
```

```python
df_patients_updated
```

*Task 9: Concatenate new billing category columns like ['InsuranceCovered', 'FinalAmount'] (column-wise).*

```python
# Create sample data for new billing category columns
num_patients = len(df_patients_updated)  # Ensure the number of rows matches df_patients_updated

# Example: 60% of BillAmount is covered by insurance for some patients, and others have a fixed deduction of 1000
insurance_covered = [df_patients_updated.loc[i, 'BillAmount'] * 0.6 if i % 2 == 0 
                     else 1000.0 for i in range(num_patients)]
final_amount = [df_patients_updated.loc[i, 'BillAmount'] - insurance_covered[i] for i in range(num_patients)]

#creating the dataframe for new billing category
new_billing_categories = pd.DataFrame({
    'InsuranceCovered': insurance_covered,
    'FinalAmount': final_amount
})
```

```python
# Concatenate the new billing category columns to patient_df_updated (column-wise)
patient_df_final = pd.concat([df_patients_updated, new_billing_categories], axis=1)
```
